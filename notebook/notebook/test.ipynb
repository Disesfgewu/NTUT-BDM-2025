```python
import h5py
import numpy as np
```

```python
filename = '../data/MillionSongSubset/A/A/A/TRAAAAW128F429D538.h5'

with h5py.File(filename, 'r') as f :
    # test = f['metadata']['songs']['audio_md5']
    song_id = f['metadata']['songs']
    # 讀取聲學特徵（例如音色）到 NumPy 數組
    segments_timbre = f['analysis']['segments_timbre'][:]
    # 打印數據形狀
    print(f"Song ID: {list(song_id)}")
    print(f"Timbre Shape: {segments_timbre.shape}")
```

```python
with h5py.File(filename, 'r') as f:
    print(f'root: {list(f.keys())}')
    metadata = f['metadata']
    print(f'metadata: {list(metadata.keys())}')
    songs = metadata['songs']
    print(f'songs: {list(songs)}')
```
